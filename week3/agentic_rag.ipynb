```python
import os
from dotenv import load_dotenv
load_dotenv()
# make sure OPENAI_API_KEY is in your .env
```

```python
import json
from langchain_core.documents import Document

with open("problem_codes.json") as f:
    codes = json.load(f)

docs = [
    Document(
        page_content=(
            f"{c['code']}: {c['category']} / {c['subcategory']} — "
            f"{c['description']} "
            f"Severity: {c['typical_severity']}. "
            f"Keywords: {', '.join(c['keywords'])}."
        ),
        metadata={"code": c["code"], "category": c["category"]}
    )
    for c in codes
]

print(f"Loaded {len(docs)} problem codes")
print(docs[0].page_content)
```

```python
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings

vectorstore = Chroma.from_documents(docs, embedding=OpenAIEmbeddings())
print("Indexed successfully")
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

results = retriever.invoke("water leaking from ceiling dripping")
for r in results:
    print(r.page_content[:100])
    print("---")
```

```python
from langchain_core.tools import tool

@tool(response_format="content_and_artifact")
def retrieve_problem_codes(query: str):
    """Search CBRE problem codes for relevant maintenance classification."""
    retrieved_docs = retriever.invoke(query)
    serialized = "\n\n".join(
        f"Source: {doc.metadata['code']}\n{doc.page_content}"
        for doc in retrieved_docs
    )
    return serialized, retrieved_docs
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model("gpt-5-mini", temperature=0)
llm_with_tools = llm.bind_tools([retrieve_problem_codes])
```

```python
from langchain_core.messages import SystemMessage

def generate_query_or_respond(state: MessagesState):
    system = SystemMessage(content=(
        "You are a CBRE maintenance call classifier. "
        "You MUST always use the retrieve_problem_codes tool to look up "
        "relevant problem codes before answering. Never respond directly."
    ))
    response = llm_with_tools.invoke([system] + state["messages"])
    return {"messages": [response]}
```

```python
from typing import Literal
from pydantic import BaseModel, Field

GRADE_PROMPT = (
    "You are a grader assessing relevance of a retrieved document to a user question.\n"
    "Here is the retrieved document:\n\n{context}\n\n"
    "Here is the user question: {question}\n"
    "If the document contains keyword(s) or semantic meaning related to the user question, "
    "grade it as relevant.\n"
    "Give a binary score 'yes' or 'no' to indicate whether the document is relevant."
)

class GradeDocuments(BaseModel):
    binary_score: str = Field(description="Relevance score: 'yes' or 'no'")

grader = llm.with_structured_output(GradeDocuments)

def grade_documents(state: MessagesState) -> Literal["generate_answer", "rewrite_question"]:
    question = state["messages"][0].content
    context = state["messages"][-1].content

    scored = grader.invoke(GRADE_PROMPT.format(context=context, question=question))
    return "generate_answer" if scored.binary_score == "yes" else "rewrite_question"
```

```python
from langchain_core.messages import HumanMessage

REWRITE_PROMPT = (
    "Look at the input and try to reason about the underlying semantic intent / meaning.\n"
    "Here is the initial question:\n"
    "---\n{question}\n---\n"
    "Formulate an improved question:"
)

def rewrite_question(state: MessagesState):
    question = state["messages"][0].content
    response = llm.invoke(REWRITE_PROMPT.format(question=question))
    return {"messages": [HumanMessage(content=response.content)]}
```

```python
from pydantic import BaseModel, Field


GENERATE_PROMPT = (
    "You are an expert CBRE building maintenance classifier.\n"
    "Use ONLY the retrieved problem codes below to classify the maintenance call.\n\n"
    "Severity guidelines:\n"
    "- Critical: Immediate danger to life or safety (gas leak, fire, trapped persons, "
    "flooding causing electrical hazard, structural collapse risk)\n"
    "- High: Significant disruption or potential for escalation (major water leak, "
    "broken security glass, HVAC failure in extreme weather, power outage)\n"
    "- Medium: Needs attention soon but not an emergency (broken door latch, minor "
    "plumbing issue, elevator malfunction without entrapment)\n"
    "- Low: Minor or cosmetic issues (flickering light, carpet stain, empty soap "
    "dispenser, aesthetic repairs)\n\n"
    "Additional rules:\n"
    "- Compound hazards: water near electrical equipment, water near server rooms, "
    "or smoke near HVAC systems should be classified as High or Critical even if "
    "the individual issues seem minor on their own.\n"
    "- Medium vs High boundary: if the issue affects multiple people, is actively "
    "worsening, or involves safety-critical equipment, escalate to High.\n"
    "- Fire doors and safety-critical equipment: any malfunction of fire doors, "
    "emergency exits, or life-safety systems is at minimum High.\n\n"
    "Retrieved problem codes:\n{context}\n\n"
    "Maintenance transcript: {question}\n\n"
    "Respond with exactly:\n"
    "- Problem code (e.g. PLUMB-001)\n"
    "- Severity (Critical / High / Medium / Low)\n"
    "- One sentence justification"
)


class MaintenanceClassification(BaseModel):
    problem_code: str = Field(description="The CBRE problem code e.g. PLUMB-001")
    severity: str = Field(description="Critical, High, Medium, or Low")
    justification: str = Field(description="One sentence explanation")

classifier = llm.with_structured_output(MaintenanceClassification)

def generate_answer(state: MessagesState):
    question = state["messages"][0].content

    # Grab all tool message content
    docs_content = "\n\n".join(
        msg.content for msg in state["messages"]
        if hasattr(msg, "type") and msg.type == "tool"
    )

    # Fallback in case tool messages aren't being captured
    if not docs_content:
        docs_content = "\n\n".join(
            msg.content for msg in state["messages"]
            if hasattr(msg, "content") and isinstance(msg.content, str) and "PLUMB" in msg.content or "HVAC" in msg.content
        )

    response = classifier.invoke(
        GENERATE_PROMPT.format(context=docs_content, question=question)
    )
    
    formatted = (
        f"Problem code: {response.problem_code}\n"
        f"Severity: {response.severity}\n"
        f"Justification: {response.justification}"
    )
    from langchain_core.messages import AIMessage
    return {"messages": [AIMessage(content=formatted)]}
```

```python
from langgraph.graph import StateGraph, START, END
from langgraph.prebuilt import ToolNode, tools_condition

workflow = StateGraph(MessagesState)

workflow.add_node("generate_query_or_respond", generate_query_or_respond)
workflow.add_node("retrieve", ToolNode([retrieve_problem_codes]))
workflow.add_node("rewrite_question", rewrite_question)
workflow.add_node("generate_answer", generate_answer)

workflow.add_edge(START, "generate_query_or_respond")
workflow.add_conditional_edges(
    "generate_query_or_respond",
    tools_condition,
    {"tools": "retrieve", END: END}
)
workflow.add_conditional_edges(
    "retrieve",
    grade_documents,
    {"generate_answer": "generate_answer", "rewrite_question": "rewrite_question"}
)
workflow.add_edge("rewrite_question", "generate_query_or_respond")
workflow.add_edge("generate_answer", END)

graph = workflow.compile()
print("Graph compiled successfully")
```

```python
from concurrent.futures import ThreadPoolExecutor, as_completed
import json, re
from langchain_core.messages import HumanMessage

with open("transcripts.json") as f:
    transcripts = json.load(f)

def run_transcript(t):
    result = graph.invoke({"messages": [HumanMessage(content=t["transcript"])]})
    answer = result["messages"][-1].content
    # Now always on its own line: "Problem code: PLUMB-001"
    match = re.search(r"Problem code:\s*([A-Z]+-\d+)", answer)
    predicted_code = match.group(1) if match else "UNKNOWN"
    return {
        "id": t["id"],
        "true_code": t["true_category"],
        "predicted_code": predicted_code,
        "true_severity": t["true_severity"],
        "correct": predicted_code == t["true_category"],
        "full_answer": answer
    }

results = [None] * len(transcripts)
with ThreadPoolExecutor(max_workers=5) as executor:
    futures = {executor.submit(run_transcript, t): i for i, t in enumerate(transcripts)}
    for future in as_completed(futures):
        i = futures[future]
        results[i] = future.result()

# Print results table
print(f"{'ID':<10} {'True':<12} {'Predicted':<12} {'Match'}")
print("-" * 45)
for r in results:
    check = "✅" if r["correct"] else "❌"
    print(f"{r['id']:<10} {r['true_code']:<12} {r['predicted_code']:<12} {check}")

accuracy = sum(r["correct"] for r in results) / len(results) * 100
print(f"\nAccuracy: {sum(r['correct'] for r in results)}/{len(results)} = {accuracy:.0f}%")

# Print wrong ones for debugging
print("\n--- Misclassified ---")
for r in results:
    if not r["correct"]:
        print(f"\n{r['id']}: true={r['true_code']}, predicted={r['predicted_code']}")
        print(r["full_answer"])
```
